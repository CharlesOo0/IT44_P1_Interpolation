## Modélisation
Modélisation d'un polynome de degré n dans la base canonique sous forme d'une liste.  
On modélise un polynôme avec une liste de n+1 éléments.  
Exemple :  
Soit $P$ un polynôme de degré 3 tel que : $$P(x) = a_0 + a_1x + a_2x + \dots + a_n$$  
Nous implémenterons ce polynôme sous python grâce à une liste de la forme :
$$P = [a_0 ,a_1 ,\dots ,a_n]$$  
On implémentera également une fonction $printPoly$ pour faciliter la lecture des polynômes

```python
def printPoly(P):
    """
    Entrées : Un Polynôme sous forme de liste
    Résultat : affichage du polynôme sous la forme a1 + a2x + ... + anx^n 
    """
    #on initialise la variable du résultat
    txt = ''
    for i,j in enumerate(P):
        #on traite l'affichage différement suivant le degré du coefficient. Les cas spéciaux sont les coefficients de degré 0 et 1
        if i!= 0 and i!= 1:
            txt += ' + ' +str(j)+'x^'+str(i)
        elif i == 1:
            txt += ' + ' +str(j)+'x'
        else:
            txt += str(j)
    #on affiche le text 
    print(txt)
```

## Évaluation
Fonction pour évaluer notre polynôme de degré n au point $t$, $t$ étant un réel.  
Pour ce faire, nous utiliseron l'algorithme de Horner pour avoir un programme plus efficient en terme de complexité.

```python
def evaluation(P,t):
    """
    Entrées : P un polynôme et t un réel
    Sorties : L'évaluation du polynôme P au rang t
    """
    #on commence par copier les coefficients du polynôme
    P_Bis = P.copy()
    #On parcourt la liste à l'envers en partant de l'avant dernier degré.
    for i in range((len(P)-2),-1,-1):
       #On applique l'algorithme de Horner
       P_Bis[i] = P[i] + t * P_Bis[i+1]
    return P_Bis[0]
```

## Addition
Fonction permettant d'additioner deux polynômes données $P_1$ et $P_2$ dans un nouveau polynôme $P_3$.  

```python
def addition(P1,P2):
    """
    Entrées : P1 et P2 deux polynômes
    Sorties : P3 un polynôme résultat de l'addition de P1 et P2
    """
    #Création de P3, résultat de l'addition de P1 et P2
    P3 = []
    #Addition des coefficients des deux polynomes tant qu'ils ont le même degré
    for i in range(min(len(P1),len(P2))):
        P3.append(P1[i]+P2[i])
    #On regarde le polynome qui n'est pas vide puis on l'ajoute dans P3
    if(len(P1)>len(P2)):
        for j in range(i+1,len(P1),1):
            P3.append(P1[j])
    else:
        for j in range(i+1,len(P2),1):
            P3.append(P2[j])
    return P3
```

## Multiplication par un réel
Fonction permettant d'addition un polynôme $P$ par un réel $r$. Le résultat, $P'$ sera un polynôme du degré de $P$ si $r \neq 0$

```python
def multiplicationReel(P,r):
    """
    Entrées : P un polynôme et r un réel
    Sorties : P2 un polynôme du rang de P résultat de la multiplication de P par r
    """
    #Création de P2 qui va stocker le résultat
    P2 = []
    #pour chaque élément dans P on ajoute à P2 le coefficient de P multiplié par le réel r
    for i in P:
        P2.append(i*r)
    return P2
```

## Multiplication de deux polynômes
Fonction permettant de multiplier deux polynômes $P_1$ et $P_2$ pour créer un nouveau polynôme $P_3$, résultat de cette multiplication.

```python
def multiplicationPoly(P1,P2):
    """
    Entrées : P1 et P2 deux polynômes
    Sorties : P3, un polynômes résultat du produit de P1 par P2
    """
    #Initialisation de la taille de P3
    P3 = [0]*(len(P1)+len(P2)-1)
    #pour chaque élément de P1
    for index1, coeff1 in enumerate(P1):
        #Pour chaque élément de P2
        for index2,coeff2 in enumerate(P2):
            #On multiplie les coefficients de P1 et P2 et on les stock dans P3
            P3[index1+index2] += coeff1*coeff2
    return P3
```

## Application aux polynômes de Lagrange

```python
import numpy as np
def lagrange(supp):
    """
    Entrées : f la fonction a interpoler, x la liste des points de l'interpolation et n le degré du polynome
    Sorties : un polynome de lagrange de degré n qui interpol f
    """
    #On utilise la fonctionnalité polynome de numpy pour récupérer un X
    X = np.poly1d([1,0])
    #P est le polynome de lagrange final
    P=0
    #On vérifie que le support est à la bonne taille
    if(len(supp[0])==len(supp[1]) and len(supp[0]) > 1 ):
        n = len(supp[0])
    else:
        return -1
    
    #Pour chaque membre du polynome cad les familles de lagrange L0, L1, ..., Ln
    for i in range(n):
        Li = 1
        for j in range(n):
            #On multiplie chaque coeeficient si i!=j
            if (i!=j):
                #Formule du cours
                Li = Li*(X-supp[0][j])/(supp[0][i]-supp[0][j])
        #On additionne le membre obtenue à P
        P += Li*supp[1][i]  
    return P
```
